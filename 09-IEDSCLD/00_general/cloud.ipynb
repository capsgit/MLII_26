## Pipeline debugging server conexion

### 🔁 Flujo completo del dibujo (lo importante)

---

#### 1️⃣ IP correct?

¿Estoy apuntando a la IP correcta?

Ejemplo:

        10.20.30.40

Si esto falla → nada funciona

---

#### 2️⃣ ping

Chequeo básico de red:

ping 10.20.30.40

👉 Pregunta:

¿el server responde?

✔️ Sí → sigue
❌ No → problema de red / firewall / server caído

---

#### 3️⃣ S W server

Aquí está la clave:

👉 Ya sabes que el server existe (ping OK)
Ahora preguntas:

¿El software del servidor está corriendo?

Ejemplos:

        tu API (FastAPI)
        un web server (nginx)
        un DB server (MySQL)

💥 Ejemplo concreto

Puedes hacer ping pero:

curl http://10.20.30.40:8000

y falla

👉 Eso significa:

        máquina viva ✅
        software muerto ❌

---

#### 4️⃣ Port check (O.S + Firewall)

Port check ↑ (OS) + Firewall

👉 Traducción:

¿el puerto está abierto?
¿el sistema operativo lo permite?
¿el firewall lo bloquea?

Ejemplo:

        puerto 8000

Problemas típicos:

        app no está escuchando
        firewall bloquea
        puerto incorrecto

---

#### 5️⃣ Whitelist / List

👉 Aquí entran reglas:

- IP permitidas
- reglas de acceso

Ejemplo:

        solo ciertas IP pueden entrar
        seguridad de red

---

#### 6️⃣ MySQL + Auth + Permissions

Última capa:

👉 Incluso si llegas al server:

necesitas login
permisos (DCL → GRANT / REVOKE)

Ejemplo:

        usuario incorrecto
        sin permisos en DB

### ⚠️ Insight

| Caso                         | Resultado       |
| ---------------------------- | --------------- |
| ping OK + server OFF         | no responde     |
| server ON + firewall bloquea | no entra        |
| todo OK + sin permisos       | acceso denegado |
