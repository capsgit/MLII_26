# 🧠 Transfer Learning mit VGG16 (Didaktisches Notebook)

Dieses Notebook erklärt Schritt für Schritt Transfer Learning – ideal für Einsteiger*innen.
## 📌 Was ist Transfer Learning?
Transfer Learning bedeutet, dass wir ein bereits trainiertes Modell wiederverwenden.

👉 Vorteil:
- weniger Daten nötig
- schnelleres Training
- oft bessere Ergebnisse

## 📦 Schritt 1: Bibliotheken importieren

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

## 🏗️ Schritt 2: VGG16 laden
Wir laden ein vortrainiertes Modell ohne Klassifikationskopf.

```python
base_model = keras.applications.VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.summary()
```

## ❄️ Schritt 3: Feature Extraction
Wir frieren alle Layer ein, damit das gelernte Wissen erhalten bleibt.

```python
base_model.trainable = False
```

## 🧩 Schritt 4: Eigenen Klassifikationskopf hinzufügen
Wir bauen ein eigenes kleines Netzwerk oben drauf.

```python
model = keras.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(1, activation='sigmoid')
])
```

## ⚙️ Schritt 5: Modell kompilieren

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

## 🔍 Schritt 6: Modellübersicht

```python
model.summary()
```

Hier würden wir das Modell trainieren.

## 🔧 Schritt 8: Fine-Tuning (Fortgeschritten)
Einige Layer werden wieder trainierbar gemacht.

```python
for layer in base_model.layers[-4:]:
    layer.trainable = True

model.compile(
    optimizer=keras.optimizers.Adam(1e-5),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```
